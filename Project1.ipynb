## Machine Learning

```python
df.head()
```

```python
#delete Loan_ID column
df.drop(columns=['Loan_ID'], inplace=True)
```

```python
df.head()
```

```python
df['Property_Area'].value_counts()
```

## Label Encoding 

```python
import warnings
warnings.filterwarnings('ignore')
df.replace({
    'Gender':{'Male': 1, 'Female': 0},
    'Married':{'Yes': 1, 'No': 0},
    'Education': {'Graduate': 1, 'Not Graduate': 0},
    'Self_Employed': {'Yes': 1, 'No': 0},
    
},inplace=True) #using dictionary method to encode label
```

```python
df.head()
```

```python
df = pd.get_dummies(df, columns=['Property_Area'], drop_first=False, dtype=int) #using dummmies method to encode labels
```

```python
df.head()
```

```python
#position loan status as last coumn
df['Loan_Status'] = df.pop('Loan_Status')
```

```python
df.head()
```

```python
#convert loan status values to 0 & 1
df.replace({'Loan_Status':{'Y':1,'N':0}}, inplace=True) #using dictionary method to encode label
```

```python
df.head()
```

## Data Standardization

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
numeric_col = ['ApplicantIncome', 'CoapplicantIncome','LoanAmount']
numeric_col
```

```python
df[numeric_col] = scaler.fit_transform(df[numeric_col])
```

```python
df.head()
```

```python
df.shape
```

```python
df['Dependents'] = df['Dependents'].astype(int)
```

```python
df['Loan_Status'].value_counts()
```

```python
df['Gender'].value_counts()
```

```python
df['Self_Employed'].value_counts()
```

```python
#replace the unknown with the most occuring value
df['Gender'] = df['Gender'].replace('Unknown', 1)
df['Self_Employed'] = df['Self_Employed'].replace('Unknown', 0)
```

```python
#converting the column datatype to interger
df['Gender'] = df['Gender'].astype(int)
df['Self_Employed'] = df['Self_Employed'].astype(int)
```

```python
df['Self_Employed'].value_counts()
```

```python
df.info()
```

Split Data -> Features & Target

```python
x = df.drop(columns=['Loan_Status']) #Features
y = df['Loan_Status'] #Target
```

```python
x
```

```python
y
```

## Test & Train Split

```python
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=5, stratify=y)
```

```python
x_train.shape, x_test.shape, y_train.shape, y_test.shape
```

## Using LogisticRegression Model

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr_model = LogisticRegression(max_iter=10000)
```

```python
lr_model
```

```python
lr_model.fit(x_train,y_train)
```

## Model Evaluation

```python
from sklearn.metrics import accuracy_score
```

```python
x_train_prediction = lr_model.predict(x_train)
x_train_accuracy = accuracy_score(y_train, x_train_prediction)
```

```python
print('Train accuracy is: ', x_train_accuracy)
```

```python
x_test_prediction = lr_model.predict(x_test)
x_test_accuracy = accuracy_score(y_test, x_test_prediction)
```

```python
print('Test accuracy is: ', x_test_accuracy)
```

```python
pickle.dump(lr_model, open('lr_model.sav', 'wb')) #write binary
```

```python
load_model = pickle.load(open('lr_model.sav', 'rb')) #read binary
```

## Building a Predictive System

```python
input_data = (1,0,0,1,0,0.072991,-0.554487,0.000000,360.0,1.0,0,0,1)

#converting input_data to a numpy array
covert_data_to_numpy = ny.array(input_data)

#In numpy conversion reshaping is usaully (1,-1)
reshaping_input_data = covert_data_to_numpy.reshape(1,-1)

prediction = load_model.predict(reshaping_input_data)
print(prediction)

if prediction == 0:
    print('Sorry you are not eligible to loan')
else:
    print('Congratulations you are eligible to loan')
```
